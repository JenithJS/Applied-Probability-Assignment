```python
import random
no_tosses=100
result= random.choices(("head","tails"), k=no_tosses)
heads=result.count("head")
tails=result.count("tails")
print(f"no of heads=",heads)
print(f"no of tails=",tails)
print(f"prob heads =", heads/no_tosses)
print(f"prob tails =", tails/no_tosses)
```

```python
import random
trails =[10, 100, 1000, 10000]
for trail in trails:
    result= random.choices(("head","tails"), k=trail)
    heads=result.count("head")
    tails=result.count("tails")
    print(f"no of trails=",trail)
    print(f"no of heads=",heads)
    print(f"no of tails=",tails)
    print(f"prob heads =", heads/trail)
    print(f"prob tails =", tails/trail)

    print()
```

```python
import random
noofrolls=10000
outcomes=[1,2,3,4,5,6]
result= random.choices(outcomes, k=noofrolls)

countof1 = result.count(1)
countof6 = result.count(6)

probof1 = countof1/noofrolls
probof6 = countof6/noofrolls

union = (sum(1 for i in result if i==1 or i==6))/noofrolls


print(f"prob of 1 =", probof1)
print(f"prob of 6 =", probof6)
print(f"union = ", union)
print(f"p(1) + p(6) = ", probof1 + probof6)

```

```python
"""For one die roll, define: A=getting an even number A={2,4,6} and B=getting a number greater than 3 B={4,5,6}  Verify: P(A∪B)=P(A)+P(B)−P(A∩B) """

import random
outcomes =[1,2,3,4,5,6]
result = random.choices(outcomes, k=1)
countofa, countofb, countofAandB, countofAorB= 0,0,0,0
ans=result[0]
if ans%2==0:
  countofa = countofa +1
if ans>3:
  countofb = countofb +1

if ans%2==0 and ans>3:
  countofAandB = countofAandB +1

if ans%2==0 or ans>3:
  countofAorB = countofAorB +1

print(f"p(a) + P(b) - P(a inter b) = ", countofa + countofb - countofAandB)
print(f"p(a union b) =", countofAorB)

print(f"result", ans)
print(f"count of a =", countofa)
print(f"count of b =", countofb)
print(f"count of AandB =", countofAandB)
```

```python
import random

num_trials = 10000
coin_outcomes = ["H", "T"]
die_outcomes = [1, 2, 3, 4, 5, 6]

countofA = 0
countofB = 0
countofAandB = 0

for i in range(num_trials):
    coin = random.choice(coin_outcomes)
    die = random.choice(die_outcomes)

    if coin == "H":
        countofA += 1

    if die % 2 == 0:
        countofB += 1

    if coin == "H" and die % 2 == 0:
        countofAandB += 1

probofA = countofA / num_trials
probofB = countofB / num_trials
probofAandB = countofAandB / num_trials

print("P(A) =", probofA)
print("P(B) =", probofB)
print("P(A intersection B) =", probofAandB)
print("P(A) * P(B) =", probofA * probofB)
```
